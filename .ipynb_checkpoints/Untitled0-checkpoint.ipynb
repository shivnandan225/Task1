```python
#data cleaning and preprocessing





df_cleaned = df.copy()
```

```python
df_cleaned.drop(columns = ['enginelocation','symboling','carheight','curbweight','stroke','compressionratio','peakrpm'],inplace=True)
```

```python
df_cleaned['drivewheel']=df_cleaned['drivewheel'].map({
    'rwd':0,
    'fwd':1,
    '4wd':2
})
```

```python
df_cleaned['fueltype']=df_cleaned['fueltype'].map({
    'gas':0,
    'diesel':1
})
```

```python
df_cleaned['aspiration']=df_cleaned['aspiration'].map({
    'std':0,
    'turbo':1
})
```

```python
df_cleaned['doornumber']=df_cleaned['doornumber'].map({
    'four': 0,
    'two':1,
})
```

```python
print(df_cleaned.columns.tolist())
```

```python
df_cleaned = pd.get_dummies(df_cleaned,columns=['cylindernumber','enginetype','fuelsystem'],drop_first=True)
```

```python
df_cleaned.head()
```

```python
df_cleaned = pd.get_dummies(df_cleaned , drop_first=True)
```

```python
df_cleaned = df_cleaned.astype(int)
```

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
cols  = ['carlength','wheelbase','carwidth','enginesize','horsepower']
```

```python
df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
```

```python
correlation_df = (
    df_cleaned.corr(numeric_only=True)['price']
      .drop('price')
      .sort_values(ascending=False)
      .reset_index()
)

correlation_df.columns = ['Feature', 'Pearson_Correlation']

correlation_df
```

```python
x = df_cleaned.drop(columns=['price'])
y = df_cleaned['price']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split( x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model1 = LinearRegression()
model1.fit(X_train,y_train)
```

```python
y_pred = model1.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
r2
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Make predictions on test data
y_pred = model1.predict(X_test)

# R² Score (commonly used as "accuracy" for Linear Regression)
r2 = r2_score(y_test, y_pred)

# Other evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

print("R² Score:", r2)
print("R² Score (%):", r2 * 100)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
```

```python
from sklearn.tree import DecisionTreeRegressor

model2 = DecisionTreeRegressor(random_state=42
                               )
model2.fit(X_train,y_train)
```

```python
y_pred_dt = model2.predict(X_test)
r2_score_dt = r2_score(y_test,y_pred_dt)
r2_score_dt
```

```python
from sklearn.ensemble import RandomForestRegressor
model3 = RandomForestRegressor(n_estimators  = 100 ,random_state = 42)
model3.fit(X_train,y_train)
```

```python
y_pred_rf = model3.predict(X_test)
r2_score_rf = r2_score(y_test,y_pred_rf)
r2_score_dt
```

```python
from sklearn.svm import SVR
model4 = SVR(kernel="rbf")
model4.fit(X_train,y_train)
```

```python
y_pred_svr = model4.predict(X_test)
r2_score_svr = r2_score(y_test,y_pred_svr)
r2_score_svr
```

```python
import pickle

# Save model
with open("model.pkl", "wb") as file:
    pickle.dump(model3, file)

# Save scaler
with open("scaler.pkl", "wb") as file:
    pickle.dump(scaler, file)

print("Model and scaler saved successfully!")
```
